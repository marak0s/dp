```python
from pathlib import Path
import sys
def find_repo_root(start: Path) -> Path:
    cur = start.resolve()
    while cur != cur.parent:
        if (cur / 'pyproject.toml').exists():
            return cur
        cur = cur.parent
    raise RuntimeError('Repo root not found')
ROOT = find_repo_root(Path.cwd())
if str(ROOT / 'src') not in sys.path:
    sys.path.insert(0, str(ROOT / 'src'))
print('ROOT=', ROOT)
```

```python
from comm_ai.experiments.run_experiment import run
from comm_ai.utils.io import load_yaml
import yaml, subprocess
cfg = load_yaml(ROOT/'src/comm_ai/config/experiments/awgn_small.yaml')
cfg['experiment']['decoders']=['bcjr']
cfg['experiment']['run_name']='nb_train_notebook'
cfg['training']['enabled']=False
tmp = ROOT/'outputs/runs/nb_train_notebook_config.yaml'
tmp.write_text(yaml.safe_dump(cfg, sort_keys=False), encoding='utf-8')
out = run(str(tmp))
subprocess.run(['python','-m','comm_ai.training.train_neural_bcjr','--config',str(tmp),'--dataset',str(out/'signals.npz')], check=True)
print('checkpoint:', out/'checkpoints/best_neural_bcjr.pt')
```
